#**Domain**

Banking and Financial Services

#**Dataset Information**

**Source:** https://github.com/ahsan084/Banking-Dataset/blob/main/Comprehensive_Banking_Database.csv

**Original Dataset Size:** 5,000 rows × 40 columns

**Dataset Used for Analysis:** 5,000 rows × 18 selected columns

**Year / Timeline:** The dataset covers the period from 2000 to 2023. Account opening records span from 2000 to 2022, while transaction, credit card, payment, and customer feedback activities are primarily recorded during 2023.

**Format:** CSV

**Dataset Column/features Description:**

* **Age** - Age of the customer.

* **Gender** - Gender of the customer.

* **City** - City associated with the customer.

* **Account Type** - Type of bank account held by the customer.

* **Account Balance** - Current balance maintained in the customer's bank account.

* **Date Of Account Opening** - Date on which the customer's bank account was opened.

* **Transaction Date** - Date on which the banking transaction occurred.

* **Transaction Type** - Type or category of banking transaction.

* **Transaction Amount** - Monetary amount involved in the transaction.

* **Loan Type** - Type of loan associated with the customer.

* **Loan Amount** - Monetary amount associated with the customer's loan.

* **Interest Rate** - Interest rate applicable to the customer's loan.

* **Loan Status** - Current status of the loan, such as approval or rejection status.

* **Card Type** - Type of credit card held by the customer.

* **Credit Limit** - Maximum credit limit available on the customer's credit card.

* **Credit Card Balance** - Outstanding balance on the customer's credit card.

* **Feedback Type** - Category or type of feedback provided by the customer.

* **Resolution Status** - Status indicating whether the customer's feedback or issue has been resolved.
