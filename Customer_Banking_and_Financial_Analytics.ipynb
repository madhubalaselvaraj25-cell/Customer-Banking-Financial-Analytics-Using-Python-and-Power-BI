```python
#  Treat Outliers
```

```python
# Numerical variables selected for outlier analysis

numeric_columns = [
    'Age',
    'Account_Balance',
    'Transaction_Amount',
    'Balance_After_Transaction',
    'Loan_Amount',
    'Interest_Rate',
    'Credit_Card_Balance'
]

# Statistical summary

df_clean[numeric_columns].describe().T
```

```python
# Check outliers using IQR method

for col in numeric_columns:

  Q1 = df_clean[col].quantile(0.25)
  Q3 = df_clean[col].quantile(0.75)
  IQR = Q3-Q1

  lower_bond = Q1 - 1.5 * IQR
  upper_bond = Q3 + 1.5 * IQR

  outliers = (
      (df_clean[col] < lower_bond)|
      (df_clean[col]> upper_bond)
  ).sum()

  print(f"\n{col}:{outliers} outliers")
```

## Outlier Detection Results

The IQR method was applied to the numerical columns: Age, Account_Balance, Transaction_Amount, Balance_After_Transaction, Loan_Amount, Interest_Rate, and Credit_Card_Balance.

No outliers were detected in any of the numerical columns. Therefore, no outlier treatment or removal was performed based on the IQR method.
